# Download and extract speech from YouTube

This file records a workflow for:

1. downloading audio from a YouTube video;
2. solving common 'yt-dlp' access problems on macOS;
3. transcribing Swedish speech automatically;
4. saving the transcript with timestamps for later flashcard creation.

The example video is a Swedish alphabet video for Children. Each letter was followed by a few typical swedish nouns that started from the letter.

---


## 1. Create and activate a Python environment

From the project folder (Ignore this step if are already inside a virtual environment:
```bash
python3 -m venv .venv
source .venv/bin/activate
```

python -m pip install --upgrade pip

## 2. Install the required tools

Install the Python packages in Terminal:

```bash
python -m pip install --upgrade yt-dlp faster-whisper
```

Install the system tools with Homebrew:

```bash
brew install ffmpeg deno
```

Check that they are available:

```bash
yt-dlp --version
ffmpeg -version
deno --version
```

### Why these tools are needed

- `yt-dlp` downloads the media.
- `ffmpeg` extracts and converts the audio.
- `deno` allows `yt-dlp` to handle YouTube's JavaScript-based checks.
- `faster-whisper` transcribes the Swedish speech.

---

## 3. Download only the audio

The following command worked on macOS with Chrome cookies:

```bash
yt-dlp \
  --cookies-from-browser chrome \
  --js-runtimes deno \
  --remote-components ejs:github \
  -x \
  --audio-format mp3 \
  -o "alphabet_audio.%(ext)s" \
  "https://www.youtube.com/watch?v=pwiyMIFLDpg"

### Explanation of the options

| Option | Meaning |
|---|---|
| `--cookies-from-browser chrome` | Uses the current Chrome login session and cookies |
| `--js-runtimes deno` | Uses Deno as the JavaScript runtime |
| `--remote-components ejs:github` | Loads the JavaScript challenge-handling component |
| `-x` | Extracts audio only |
| `--audio-format mp3` | Converts the audio to MP3 |
| `-o "alphabet_audio.%(ext)s"` | Saves the result as `alphabet_audio.mp3` |

### macOS Keychain message

When Chrome cookies are used, macOS may display a message such as:

> security wants to access key "Chrome Safe Storage" in your keychain.

Enter your Mac login password and choose **Allow**.

## 4. Confirm that the audio file exists

````python
from pathlib import Path

audio_path = Path('alphabet_audio.mp3')

if not audio_path.is_file():
    raise FileNotFoundError(f"Audio file not found: {audio_path.resolve()}")

print("Audio file:", audio_path.resolve())
print("Size:", round(audio_path.stat().st_size / 1_000_000, 2), "MB")
````

## 5. Transcribe the audio speech with Faster Whisper

Install the package if its has not already been installed:

````python
# pip install faster-whisper
from faster_whisper import WhisperModel
model = WhisperModel(
    model_size_or_path = "medium",
    device = "cpu",
    compute_type = "int8"
)

segments, info = model.transcribe(
    "alphabet_audio.mp3",
    language="sv",
    beam_size=5,
    word_timestamps=True,
)

print("Detected language:", info.language)
print("Language probability:", info.language_probability)
````

### Notes

To check the parameters of the model, one can use '?WhisperModel', '?model.transcribe' to check the function's other inputs.

- `language="sv"` tells Whisper that the speech is Swedish.
- `word_timestamps=True` returns approximate start and end times for each word.
- `compute_type="int8"` is suitable for CPU use and reduces memory consumption.
- A smaller model, such as `"small"` or `"medium"`, can be used when speed is more important than accuracy.

## 6. Print the transcript with timestamps

````python
transcript_rows = []

for segment in segments:
    text = segment.text.strip()

    transcript_rows.append(
        {
            "start": segment.start,
            "end": segment.end,
            "text": text,
        }
    )

    print(f"[{segment.start:7.2f}–{segment.end:7.2f}] {text}")
````

### Extract the information we need for Swedish flashcard creation

````python
import pandas as pd

transcript_df = pd.DataFrame(transcript_rows)

swedish_df = transcript_df["text"].str.extract(
    r"^\s*([A-ZÅÄÖ])\s+som\s+i\s+([A-Za-zÅÄÖåäö-]+)",
    expand=True,
)

swedish_df.columns = ["letter", "word"]
swedish_df = swedish_df.dropna().reset_index(drop=True)

swedish_df.head()
````

### How the regular expression works

```python
r"^\s*([A-ZÅÄÖ])\s+som\s+i\s+([A-Za-zÅÄÖåäö-]+)"
```

- ^ = beginning of the text
- \s* = zero or more spaces
- ([A-ZÅÄÖ]) = capture one uppercase Swedish letter
- \s+ = one or more spaces
- som\s+i = match the words som i
- ([A-Za-zÅÄÖåäö-]+) = capture the example word starting with this letter


## 7. Save the results

````python
swedish_df.to_csv(
    "alphabet_swedish.csv",
    index=False,
    sep=";",
    encoding="utf-8-sig",
)
````
