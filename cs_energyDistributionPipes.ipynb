### Average Flow Velocity

The average flow velocity in each pipe was calculated using

$$
U=\frac{Q}{A}
$$

where \(Q\) is the flow rate and \(A\) is the cross-sectional area of the pipe. For a circular pipe,

$$
A=\frac{\pi D^2}{4}
$$

and therefore

$$
U=\frac{4Q}{\pi D^2}.
$$

The calculated velocities for Pipes 1–3 for both experiments are presented in Table X.

```python
import numpy as np
import pandas as pd

# Constants
nu = 1.004e-6      # kinematic viscosity of water, m2/s
k = 0.0015e-3      # PVC roughness, m

# Mean flow rates from your measured data
Q_A = df[df['Experiment'] == 'A']['Q (m3/s)'].mean()
Q_B = df[df['Experiment'] == 'B']['Q (m3/s)'].mean()

pipe_df = pd.DataFrame({
    'Pipe': ['Pipe 1', 'Pipe 2', 'Pipe 3'],
    'D_H (m)': [0.080, 0.019, 0.040]
})
pipe_df['A (m2)'] = np.pi * pipe_df['D_H (m)']**2 / 4

pipe_df['U_A (m/s)'] = Q_A / pipe_df['A (m2)']
pipe_df['U_B (m/s)'] = Q_B / pipe_df['A (m2)']

pipe_df['k/D_H'] = k / pipe_df['D_H (m)']

pipe_df['Re_A'] = pipe_df['U_A (m/s)'] * pipe_df['D_H (m)'] / nu
pipe_df['Re_B'] = pipe_df['U_B (m/s)'] * pipe_df['D_H (m)'] / nu

pipe_df.round(5)
```

```python
import pandas as pd
import numpy as np

# Constants
g = 9.81

# Manometer data
df_h = pd.DataFrame({
    'Point': [1, 2, 3, 4, 5, 6, 7, 8],
    'A: h (m)': [0.213, 0.213, 0.213, 0.102, 0.054, 0.050, 0.073, 0.073],
    'B: h (m)': [0.203, 0.203, 0.203, 0.098, 0.054, 0.050, 0.070, 0.071]
}).set_index('Point')

# Pipe section data
friction_df = pd.DataFrame({
    'Pipe section': ['Pipe 1', 'Pipe 2', 'Pipe 3'],
    'Manometers': ['2–3', '4–5', '6–7'],
    'Upstream point': [2, 4, 6],
    'Downstream point': [3, 5, 7],
    'L (m)': [0.210, 0.670, 0.292],
    'D_H (m)': [0.080, 0.019, 0.040],
    'U_A (m/s)': [0.06338, 1.12361, 0.25352],
    'U_B (m/s)': [0.06064, 1.07506, 0.24256]
})

# Head loss from energy equation: hf = h1 - h2
friction_df['h_f A (m)'] = (
    friction_df['Upstream point'].map(df_h['A: h (m)'])
    - friction_df['Downstream point'].map(df_h['A: h (m)'])
)

friction_df['h_f B (m)'] = (
    friction_df['Upstream point'].map(df_h['B: h (m)'])
    - friction_df['Downstream point'].map(df_h['B: h (m)'])
)

# Experimental Darcy friction factor:
# hf = f * (L / D_H) * (U^2 / 2g)
# f = hf * D_H / L * 2g / U^2
friction_df['f_A'] = (
    friction_df['h_f A (m)']
    * friction_df['D_H (m)']
    / friction_df['L (m)']
    * (2 * g)
    / friction_df['U_A (m/s)']**2
)

friction_df['f_B'] = (
    friction_df['h_f B (m)']
    * friction_df['D_H (m)']
    / friction_df['L (m)']
    * (2 * g)
    / friction_df['U_B (m/s)']**2
)

# Final table
friction_table = friction_df[
    ['Pipe section', 'Manometers', 'L (m)', 'h_f A (m)', 'f_A', 'h_f B (m)', 'f_B']
].round({
    'L (m)': 3,
    'h_f A (m)': 3,
    'f_A': 4,
    'h_f B (m)': 3,
    'f_B': 4
})

friction_table
```

```python
import pandas as pd

g = 9.81

df_h = pd.DataFrame({
    'Point': [1, 2, 3, 4, 5, 6, 7, 8],
    'A: h (m)': [0.213, 0.213, 0.213, 0.102, 0.054, 0.050, 0.073, 0.073],
    'B: h (m)': [0.203, 0.203, 0.203, 0.098, 0.054, 0.050, 0.070, 0.071]
}).set_index('Point')

U_A = {'Reservoir': 0, 'Pipe 1': 0.06338, 'Pipe 2': 1.12361, 'Pipe 3': 0.25352}
U_B = {'Reservoir': 0, 'Pipe 1': 0.06064, 'Pipe 2': 1.07506, 'Pipe 3': 0.24256}

components = pd.DataFrame({
    'Component': ['Entrance', 'Contraction', 'Expansion', 'Exit'],
    'Manometers': ['1–2', '3–4', '5–6', '7–8'],
    'Up point': [1, 3, 5, 7],
    'Down point': [2, 4, 6, 8],
    'Up section': ['Reservoir', 'Pipe 1', 'Pipe 2', 'Pipe 3'],
    'Down section': ['Pipe 1', 'Pipe 2', 'Pipe 3', 'Reservoir']
})

def calculate_ht_kt(row, exp):
    h_col = f'{exp}: h (m)'
    U = U_A if exp == 'A' else U_B

    h_up = df_h.loc[row['Up point'], h_col]
    h_down = df_h.loc[row['Down point'], h_col]

    U_up = U[row['Up section']]
    U_down = U[row['Down section']]

    ht = (h_up + U_up**2/(2*g)) - (h_down + U_down**2/(2*g))

    U_ref = max(U_up, U_down)
    kt = ht / (U_ref**2/(2*g))

    return pd.Series([ht, U_ref, kt])

components[['h_t A (m)', 'U_ref A (m/s)', 'k_t A']] = components.apply(
    calculate_ht_kt, axis=1, exp='A'
)

components[['h_t B (m)', 'U_ref B (m/s)', 'k_t B']] = components.apply(
    calculate_ht_kt, axis=1, exp='B'
)

table6 = components[
    ['Component', 'Manometers', 'h_t A (m)', 'k_t A', 'h_t B (m)', 'k_t B']
].round(4)

table6
```
